```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Latihan 2
import pandas as pd
df = pd.read_csv('penjualan_nusantaramart - penjualan_nusantaramart.csv')
total_kategori = df.groupby('kategori_produk')['total'].sum().sort_values(ascending=False)
print(total_kategori)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=total_kategori.values, y=total_kategori.index, palette='Blues_r') plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()
```

1. Berdasarkan hasil visualisasi Bar Chart, kategori produk dengan total penjualan tertinggi adalah Fashion, yaitu sebesar Rp17.695.000. Sedangkan kategori dengan total penjualan terendah adalah Makanan & Minuman, yaitu sebesar Rp3.029.000.
2.  Berdasarkan hasil grafik, manajer Nusantara Mart disarankan untuk mempertahankan dan meningkatkan stok produk Fashion dan Elektronik karena kedua kategori tersebut memiliki total penjualan tertinggi. Sementara itu, kategori Makanan & Minuman perlu mendapatkan perhatian lebih melalui promosi, diskon, atau strategi pemasaran agar penjualannya dapat meningkat.

```python
#Latihan 3
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.to_period('M')
tren_bulanan = df.groupby('bulan')['total'].sum()
plt.figure(figsize=(8,5))
plt.plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

1. Bulan apa yang memiliki penjualan tertinggi? Menurutmu, apa kemungkinan penyebabnya?

Jawaban: Bulan dengan penjualan tertinggi adalah bulan yang memiliki total penjualan paling besar pada grafik. Kemungkinan penyebabnya adalah meningkatnya minat pelanggan, adanya promo atau diskon, serta kebutuhan pelanggan terhadap produk tertentu.
2. Apakah tren penjualan cenderung naik, turun, atau fluktuatif? Jelaskan.

Jawaban: Tren penjualan dapat dikatakan fluktuatif apabila grafik mengalami kenaikan dan penurunan dari bulan ke bulan. Hal ini menunjukkan bahwa jumlah penjualan Nusantara Mart belum stabil.

```python
#Latihan 4
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x='jumlah', y='total', hue='kategori_produk')
plt.title('Hubungan Jumlah Pembelian dengan Total Nilai Transaksi')
plt.xlabel('Jumlah Barang')
plt.ylabel('Total (Rp)')
plt.tight_layout()
plt.show()
```

1. Apakah terlihat pola hubungan antara jumlah barang dan total nilai transaksi? Jelaskan hubungan tersebut (positif/negatif/tidak berhubungan).

Jawaban: Jika titik-titik pada grafik cenderung bergerak naik dari kiri ke kanan, berarti terdapat hubungan positif. Artinya, semakin banyak jumlah barang yang dibeli, biasanya semakin besar total nilai transaksinya.
2. Apakah ada titik data yang terlihat jauh dari pola umum (kemungkinan outlier)? Sebutkan kategorinya jika terlihat dari warna.

Jawaban: Titik yang letaknya jauh dari kumpulan titik lainnya dapat dianggap sebagai kemungkinan outlier. Kategori produk dapat diketahui melalui warna titik pada grafik.

```python
#Latihan 5
# Histogram distribusi rating pelanggan
plt.figure(figsize=(7,5))
sns.histplot(df['rating_pelanggan'], bins=5, color='seagreen')
plt.title('Distribusi Rating Pelanggan')
plt.xlabel('Rating')
plt.ylabel('Frekuensi')
plt.show()
# Boxplot jumlah pembelian per kategori (deteksi outlier)
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='kategori_produk', y='jumlah')
plt.title('Sebaran Jumlah Pembelian per Kategori Produk')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

1. Bagaimana bentuk distribusi rating pelanggan? Apakah cenderung tinggi, rendah, atau merata?

Jawaban: Distribusi rating pelanggan dapat diketahui dari histogram. Jika sebagian besar data berada pada rating 4–5, berarti pelanggan cenderung memberikan penilaian yang tinggi. Hal ini menunjukkan bahwa sebagian besar pelanggan merasa puas terhadap produk atau pelayanan.
2. Kategori produk mana yang memiliki titik-titik di luar kotak boxplot (outlier)? Apa kemungkinan penyebabnya?

Jawaban: Kategori produk yang memiliki titik di luar kotak boxplot menunjukkan adanya jumlah pembelian yang berbeda jauh dari sebagian besar transaksi. Hal ini kemungkinan disebabkan oleh pembelian dalam jumlah besar, misalnya pembelian oleh reseller atau pelanggan yang membeli produk untuk dijual kembali.

```python
#Latihan 6
fig, axes = plt.subplots(2, 2, figsize=(13,9))

sns.barplot(x=total_kategori.values, y=total_kategori.index, ax=axes[0,0])
axes[0,0].set_title('Total Penjualan per Kategori')

axes[0,1].plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o')
axes[0,1].set_title('Tren Penjualan Bulanan')

sns.histplot(df['rating_pelanggan'], bins=5, ax=axes[1,0])
axes[1,0].set_title('Distribusi Rating Pelanggan')

sns.boxplot(data=df, x='kategori_produk', y='jumlah', ax=axes[1,1])
axes[1,1].set_title('Sebaran Jumlah per Kategori')

plt.tight_layout()
plt.show()
```
